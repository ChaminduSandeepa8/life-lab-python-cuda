```python
from pathlib import Path
import os, sys, subprocess, json
PROJECT = Path.cwd()
if not (PROJECT / 'life').is_dir():
    PROJECT = PROJECT.parent if (PROJECT.parent / 'life').is_dir() else PROJECT / 'life-lab-python-cuda'
assert (PROJECT / 'life').is_dir(), 'Set PROJECT to the clone containing app.py and life/'
os.chdir(PROJECT)
if str(PROJECT) not in sys.path: sys.path.insert(0, str(PROJECT))
print('Project:', PROJECT)
```
